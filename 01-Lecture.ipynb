<h2><strong>What is Deep Learning?</strong></h2>

Deep Learning is a subset of Machine Learning that teaches computers to do what comes naturally to humans: learn by example. It uses algorithms called <strong style="color:green;">Artificial Neural Networks</strong> that are inspired by the structure and function of the human brain to process data, recognize patters, and make decision.

<h2><strong> How it is different from ML and AI</strong></h2>
<strong>Artificial Intelligence</strong>: The broadcast concept. It is any technique that enables computers to mimic human intelligence (eg. a smiple chess bot).

<strong>Machine learning</strong>: A subset of AI. Instead of giving the computer explicit rules, we give it data and  let it learn the rules (eg. predicting house price)

<strong>Deep learning</strong>: A specialized subset of ML. It uses complex, multi layered neural networks to solve highly complex problems without human intervention in feature extracation.

<h2><strong>Why Deep learning is popular today ?</strong></h2>

1. Massive Data: We generate exabytes of data daily (images ,text, sensors...). Deep learning models thrive on huge amount of data.

2. Computing power: Modern GPUs , originally designed for video games, are perfect for the heavy matrix math required by the Deep Learning.

3. Algorithmic Breakthrough: Better optimization techniques and activation functions have made it easier to train networks with many layers(the "deep" in deep learning).

<h2><strong>What is Neural Network</strong></h2>

A neural network is a computing system made up of interconnected nodes that work together to process information and produce output.

<h2><strong>Basic Intution</strong></h2>

In your brain, you have billions of cells called neurons. When you touch something hot, neurons send electrical signals to each other, eventually telling you hand to pull away. Artificial neural network mimic this. they have neurons that receive input data, process it, and pass signals to other neurons to predict and outcome.

<h2><strong>Key components</strong></h2>

- Neurons(nodes): The basic mathematical units that hold a number.
- Weights: The "strength" or "importance" of a connection between two neurons. if an input is highly relevant , its weight will be larger.
- Bias: An extra constant number added to the neuron to shift the activation function, acting like a basline threshold. It ensures the network can model patterns even when all inputs are zero.
- Activation Function: the mathematical "filter" or "gatekeeper" at the end of a neuron. It decides whether the neuron's signal is strong enough to be passed on to next layer.

<h2><strong>Types of neural networks</strong></h2>

- ANN (Artificial neural Network)
- CNN (Convolutional Neural network)
- RNN (Recurrent Neural Network)

<h2><strong>ANN</strong></h2>

- What it is: The standard, most basic type of neural network where neurons are arranged in layer, and every neuron in one layer connects to every neuron in the next (fully connected)
- Where it is used: Tabular data (row and columns)
- Real life usecases: prediction, fraud detection , estimating values.

<h2><strong>CNN</strong></h2>

- what it is: A network specifically designed to process grid like data, It uses filters to scan across data in chunks, identifying spatial patterns like edges, shapes, and textures
- where it is used: Visual data(2D grids of pixels.)
- Real life example: Facial recognition, self driving cars, X-Rays

<h2><strong>RNN</strong></h2>

- What it is: A network with "memory", Unlike ANNs, RNNs loop information back into themselves, allowing them to understand the context of what happend previously in a sequence.
- Where it is used: Sequential or time-series data
- real life usecases: Predicting stock, translating languages, speech to text features.

<center><h2 style="color:magenta;"><strong>Perceptron</strong></h2></center>

The perceptron is the building block of all neural network. it is the simplest type of artificial neuron. Think of it as a tiny decision-making machine that takes in multiple inputs, wighs their importance, and spits out a simple yes/no (binary) answer.

1. <code>Weighted Sum:</code> We multiply each input by its weight and add them all together, then add the bias.

$$ z = (x_1.w_1) + (x_2.w_2)+...+(x_n.w_n)+b $$

2. <code>Activation Function: </code> We pass the sum z through activation function , if z >=0, the output is 1, if z   < 0 ,output is 0

<h2><strong>Types of perceptron</strong></h2>

- Single layer
- Multi layer

<h3><strong>Single layer perceptron(SLP)</strong></h3>

It consists of just two parts: an input layer and an output Node, There are no hidden layer in between.

<code>Steps</code>

1. Receive the raw data inputs.
2. Multiply each input by its specific weight
3. Sum these values together and add the bias
4. Apply a strict activation function
5. Produce a single, binary output

<h3><strong>Multi layer perceptron</strong></h3>

This the true deep neural network. It consists of an input layer, one or more hidden layers (layers of neurons between input and output), and an output layer.

1. Data enters the input layer.
2. Data flows into the first hidden layer. Every neuron in the hidden layer performs the weigted sum and activation method
3. The outputs of the first hidden layer become the inputs for the next hidden layer.
4. This process continues until the data reaches the output layer
5. the output layer calculates the final prediction

```python
import numpy as np

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([0,0,0,1])


weights = np.zeros(X.shape[1])

bias = 0

learning_rate = 0.1

e = 10

def step_function(z):
    return 1 if z>=0 else 0


for e_ in range(e):
    for i in range(len(X)):
        linear_output = np.dot(X[i],weights)+bias
        y_pred = step_function(linear_output)
        error = y[i] - y_pred

        weights = weights + learning_rate * error * X[i]

        bias = bias + learning_rate * error

        print(f"Input {X[i]},Prediction {y_pred} error {error}")


print(weights)
print(bias)
```

```python
for i in range(len(X)):
    output = np.dot(X[i],weights) + bias
    prediction = step_function(output)
    print(f"Input {X[i]}----> Prediction {prediction}")
```
